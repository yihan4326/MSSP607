```python
from google.colab import userdata
import os
import subprocess
import pandas as pd

# Go back to the main Colab directory
os.chdir('/content')

# Get GitHub token from Colab Secrets
github_token = userdata.get('Git_Key')

# Set the GitHub repository owner and name
owner = 'yihan4326'
repository = 'MSSP607'

# Create the clone URL
clone_url = f'https://{github_token}@github.com/{owner}/{repository}.git'

# Clone the repository if it does not already exist
if not os.path.exists(repository):
    subprocess.run(['git', 'clone', clone_url], check=True)

# Change directory to the repository
os.chdir(repository)

print(f"Changed directory to: {os.getcwd()}")
```

```python
# Read the Student Performance Data Set
df = pd.read_csv('data/Student Performance Data.csv')

# Check the size of the data
print("Data shape:", df.shape)

# Display the first five rows
df.head()
```

```python
# Create a lambda function to calculate the average score
new_df = (lambda data: data.assign(
    **{'average score': (data['math score'] + data['reading score'] + data['writing score']) / 3}
))(df)

# Display the new DataFrame
new_df.head()
```
