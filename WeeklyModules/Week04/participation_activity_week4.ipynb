```python
# read parsed.csv
import pandas as pd
earthquakes = pd.read_csv(data_path)
earthquakes.head()
```

```python
# a.the 95th percentile of earthquake magnitude in Japan using magType of "mb"
japan_mb = earthquakes[(earthquakes['parsed_place'] == 'Japan') & (earthquakes['magType'] == 'mb')]
japan_mb['mag'].quantile(0.95)
```

```python
# b.the percentage of earthquakes in Indonesia that were coupled with tsunamis
indonesia = earthquakes[earthquakes['parsed_place'] == 'Indonesia']
indonesia_tsunamis = indonesia[indonesia['tsunami'] == 1]
percentage = len(indonesia_tsunamis) / len(indonesia) * 100
print(percentage)
```

```python
# c.summary statistics for earthquakes in Nevada
nevada = earthquakes[earthquakes['parsed_place'] == 'Nevada']
nevada.describe()
```

```python
# d. add the column of ring of fire
import re
ring_of_fire = r'Bolivia|Chile|Ecuador|Peru|Costa Rica|Guatemala|(?<!New )Mexico|Japan|Philippines|Indonesia|New Zealand|Antarctic|Canada|Fiji|Alaska|Washington|California|Russia|Taiwan|Tonga|Kermadec Islands'
earthquakes['ring_of_fire'] = earthquakes['parsed_place'].str.contains(ring_of_fire,regex=True,na=False)
earthquakes[['parsed_place', 'ring_of_fire']].head()
```

```python
# e. count earthquakes inside and outside the ring of fire
earthquakes['ring_of_fire'].value_counts()
```

```python
# f. the tsumi count along the ring of fire
ring = earthquakes[earthquakes['ring_of_fire'] == True]
ring_tsunamis = ring[ring['tsunami'] == 1]
tsunami_count = len(ring_tsunamis)
print(tsunami_count)
```
