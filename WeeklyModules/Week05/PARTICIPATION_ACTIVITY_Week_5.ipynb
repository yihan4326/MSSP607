```python
# Read each file into a dataframe
aapl = pd.read_csv(data_folder / 'aapl.csv')
amzn = pd.read_csv(data_folder / 'amzn.csv')
nflx = pd.read_csv(data_folder / 'nflx.csv')
fb = pd.read_csv(data_folder / 'fb.csv')
goog = pd.read_csv(data_folder / 'goog.csv')
```

```python
# Add ticker column
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
nflx['ticker'] = 'NFLX'
fb['ticker'] = 'FB'
goog['ticker'] = 'GOOG'
```

```python
# create a single dataframe
faang = pd.concat([fb, aapl, amzn, nflx, goog])
faang.head()
```

```python
# Save faang.csv
faang.to_csv(data_folder / 'faang.csv', index=False)
```

```python
# Read faang.csv and change data types
faang = pd.read_csv(data_folder / 'faang.csv')
faang['date'] = pd.to_datetime(faang['date'])
faang['volume'] = faang['volume'].astype(int)
faang = faang.sort_values(['date', 'ticker'])
faang.head()
```

```python
# The 7 rows with highest volume
faang.sort_values('volume', ascending=False).head(7)
print(faang.sort_values('volume', ascending=False).head(7))
```

```python
# Use melt()
faang_long = faang.melt(id_vars=['date', 'ticker'])
faang_long.head()
```
