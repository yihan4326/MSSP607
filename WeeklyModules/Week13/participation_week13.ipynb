
In the context of downloading data from Yahoo Finance using the yfinance library, the period parameter defines the range of data to download and can have the following values:

*    1d, 5d: data for the past 1 or 5 days

*   1mo, 3mo, 6mo: data for the past 1, 3, or 6 months
*   1y, 2y, 5y, 10y, ytd: data for the past 1, 2, 5, or 10 years, or year to date.


*   max: data for the maximum available period.


These values allow you to specify the time frame for which you want to retrieve historical data for a given stock.

#Task 1: Read into seperate data frame the following tickers: Apple, Facebook, Google, and Netflix; use Yahoo finance.

```python
tickers = {'Apple': 'AAPL', 'Facebook': 'META', 'Google': 'GOOGL', 'Netflix': 'NFLX'}
start_date = '2023-01-01'
end_date = '2024-01-01'

apple = yf.Ticker('AAPL').history(start=start_date, end=end_date)
facebook = yf.Ticker('META').history(start=start_date, end=end_date)
google = yf.Ticker('GOOGL').history(start=start_date, end=end_date)
netflix = yf.Ticker('NFLX').history(start=start_date, end=end_date)

for name, data in zip(tickers, [apple, facebook, google, netflix]):
    print(name, data.shape)
    display(data.head())
```

#Task 2:  Join the individual stock data frames into a single data frame.

```python
combined = pd.concat({'AAPL': apple, 'META': facebook, 'GOOGL': google, 'NFLX': netflix}, names=['Ticker', 'Date'])
combined = combined.reset_index()
combined.head()
```

# Task 3: Create a visualization of the following financial metrics for each stock:  quickRatio, currentRatio, totalRevenue, debtToEquity, revenuePerShare, returnOnAssets, returnOnEquity, grossProfits


```python
metrics = ['quickRatio', 'currentRatio', 'totalRevenue', 'debtToEquity', 'revenuePerShare', 'returnOnAssets', 'returnOnEquity', 'grossProfits']
financial_data = {}

for name, ticker in tickers.items():
    info = yf.Ticker(ticker).info
    financial_data[name] = {metric: info.get(metric, np.nan) for metric in metrics}

financial_df = pd.DataFrame(financial_data)
display(financial_df)

fig, axes = plt.subplots(2, 4, figsize=(18, 8))
for ax, metric in zip(axes.flat, metrics):
    financial_df.loc[metric].plot(kind='bar', ax=ax, title=metric)
    ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

#Task 4: Create a single boxplot for the Closing price for each stock.


```python
closing_prices = pd.DataFrame({
    'Apple': apple['Close'],
    'Facebook': facebook['Close'],
    'Google': google['Close'],
    'Netflix': netflix['Close']
})

closing_prices.plot(kind='box', figsize=(9, 6))
plt.title('Closing Prices in 2023')
plt.ylabel('Closing Price (USD)')
plt.tight_layout()
plt.show()
```

# Task 5: Create a daily trading volume visualization for the year 2023 for each stock.


```python
fig, axes = plt.subplots(4, 1, figsize=(12, 12), sharex=True)
for ax, (name, data) in zip(axes, zip(tickers, [apple, facebook, google, netflix])):
    data['Volume'].plot(ax=ax, title=f'{name} Daily Trading Volume in 2023')
    ax.set_ylabel('Volume')
plt.tight_layout()
plt.show()
```
