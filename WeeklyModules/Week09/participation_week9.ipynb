```python
# Import libraries
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

# Choose an image category
category = 'Cats'

# Set the webpage URL
url = 'https://commons.wikimedia.org/wiki/Category:Cats'

# Add request headers
headers = {
    'User-Agent': 'MSSP607StudentImageDownloader/1.0 (educational project)'
}

# Request the webpage
response = requests.get(url, headers=headers, timeout=30)

# Check the response status
print("Status code:", response.status_code)

# Check for errors
response.raise_for_status()

# Parse the webpage
soup = BeautifulSoup(response.text, 'html.parser')

# Find image tags
images = soup.find_all('img')

print("Number of image tags found:", len(images))
```

```python
# Find images in the category gallery
gallery = soup.find_all('div', class_='gallerytext')

# Create a list for image URLs
image_urls = []

# Find image links
for image in soup.select('.gallerybox img'):
    image_url = image.get('src')

    if image_url:
        full_url = urljoin(response.url, image_url)
        image_urls.append(full_url)

# Remove duplicate URLs
image_urls = list(dict.fromkeys(image_urls))

# Display the results
print("Number of cat images found:", len(image_urls))
print(image_urls[:5])
```

```python
# Set the folder for downloaded images
folder = '/content/MSSP607/WeeklyModules/Week09/cats_images'

# Create the folder
os.makedirs(folder, exist_ok=True)
# Create a folder for downloaded images
folder = 'WeeklyModules/Week09/cats_images'
os.makedirs(folder, exist_ok=True)

# Download all found images
downloaded = 0

for i, image_url in enumerate(image_urls, start=1):

    try:
        # Request the image
        res = requests.get(image_url, headers=headers, timeout=30)

        # Check for errors
        res.raise_for_status()

        # Check the image format
        image_type = res.headers.get('Content-Type', '').split(';')[0]

        extensions = {
            'image/jpeg': '.jpg',
            'image/png': '.png',
            'image/webp': '.webp',
            'image/gif': '.gif'
        }

        if image_type not in extensions:
            print("Skipped unsupported image:", i)
            continue

        # Create a file name
        filename = f'cat_{i:02d}' + extensions[image_type]
        file_path = os.path.join(folder, filename)

        # Save the image
        with open(file_path, 'wb') as image_file:
            for chunk in res.iter_content(100000):
                image_file.write(chunk)

        downloaded += 1
        print("Downloaded:", filename)

    except requests.RequestException as error:
        print("Download failed:", error)

print("Total images downloaded:", downloaded)
```
