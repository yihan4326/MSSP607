## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
mean=sum(salaries)/len(salaries)
print(mean)
```

```python
import statistics
mean=statistics.mean(salaries)
print(mean)
```

### median

```python
sorted_salaries=sorted(salaries)
n=len(sorted_salaries)
median=(sorted_salaries[n//2-1]+sorted_salaries[n//2])/2
print(median)
```

```python
median=statistics.median(salaries)
print(median)
```

### mode

```python
mode=max(salaries,key=salaries.count)
print(mode)
```

```python
mode=statistics.mode(salaries)
print(mode)
```

### sample variance
Remember to use Bessel's correction.

```python
sample_variance=sum((x-mean)**2 for x in salaries)/(len(salaries)-1)
print(sample_variance)
```

```python
sample_variance=statistics.variance(salaries)
print(sample_variance)
```

### sample standard deviation
Remember to use Bessel's correction.

```python
standard_deviation=sample_variance**0.5
print(standard_deviation)
```

```python
standard_deviation=statistics.stdev(salaries)
print(standard_deviation)
```

## Exercise 6: Calculating more statistics
### range

```python
range=max(salaries)-min(salaries)
print(range)
```

### coefficient of variation

```python
coefficient_of_variation=standard_deviation/mean
print(coefficient_of_variation)
```

### interquartile range

```python
Q1=statistics.quantiles(salaries)[0]
Q3=statistics.quantiles(salaries)[2]
IQR=Q3-Q1
print(IQR)
```

### quartile coefficent of dispersion

```python
quartile_coefficient=(Q3-Q1)/(Q3+Q1)
print(quartile_coefficient)
```

## Exercise 7: Scaling data
### min-max scaling

```python
min_max_scaled=[(x-min(salaries))/(max(salaries)-min(salaries))
for x in salaries]
print(min_max_scaled)
```

### standardizing

```python
standardized=[(x-mean)/standard_deviation
for x in salaries]
print(standardized)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
import numpy as np
np.cov(min_max_scaled,standardized)
```

### Pearson correlation coefficient ($\rho$)

```python
from scipy.stats import pearsonr
r=pearsonr(min_max_scaled,standardized)
print(r)
```
